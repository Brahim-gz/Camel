```python
from bs4 import BeautifulSoup
import requests
import time
from selenium import webdriver

browser = webdriver.Chrome()
browser.get('https://www.palestine-studies.org/en/blogs/explorer?f%5B0%5D=field_blog_series%3A19536')
ScrollNumber = 20
for i in range(1,ScrollNumber):
    browser.execute_script("window.scrollTo(1,50000)")
    time.sleep(5)
```

```python
blogs = []
soup = BeautifulSoup(browser.page_source, "html.parser")
elements = soup.find_all('div', class_ = 'explorerClass')
id = 0
for element in elements:
        title = element.find('div', class_ = 'card-bodytext blogtext').text
        blog_page = element.find('div', class_ = 'card-bodytext blogtext').a["href"]
        html_text = requests.get('https://www.palestine-studies.org'+blog_page).text
        soup = BeautifulSoup(html_text, 'lxml')
        content = soup.find('div', class_ = 'blog-body').text.replace('\n',' ').replace("\xa0","").replace("\\'","'")
        author = soup.find('div', class_ = 'row blog-info-grid').a.text
        date = soup.find('div', class_ = 'row blog-info-grid').span.text
        
        L = [ '> Blog title : '+title+'\n', '> Author : '+author+'\n', '> Date : '+date+'\n', '> Content : \n'+content]
        id+=1
        if(id>90):
            filename = 'Blog-'+str(id)+'.txt'
            with open(filename, 'w', encoding='utf-8') as file:
                file.writelines(L)
            file.close()

        blogs.append({
            "title":title,
            "author":author,
            "date":date,
            "content":content
                       
        })
```
